```python

import numpy as np


def simulate_lidar_readings(
    number_of_rays=9,
    max_range=5.0,
    obstacle_distance=2.0
):
    """
    Generate example LiDAR readings for initial testing.

    These are synthetic readings, not real Webots measurements.
    """

    if number_of_rays < 1:
        raise ValueError("number_of_rays must be at least 1.")

    if max_range <= 0:
        raise ValueError("max_range must be positive.")

    if not 0 <= obstacle_distance <= max_range:
        raise ValueError(
            "obstacle_distance must be between 0 and max_range."
        )

    readings = np.full(number_of_rays, max_range, dtype=float)

    # Simulate an obstacle in the centre of the sensor's view.
    centre_ray = number_of_rays // 2
    readings[centre_ray] = obstacle_distance

    return readings


if __name__ == "__main__":
    distances = simulate_lidar_readings()

    print("Simulated LiDAR distances (metres):")
    print(distances)

    print("Closest obstacle (metres):", np.min(distances))
```

```python

import numpy as np

def simulate_lidar_readings(
    number_of_rays=9,
    max_range=5.0,
    obstacle_distance=2.0
):
    if number_of_rays < 1:
        raise ValueError("number_of_rays must be at least 1")
    if max_range <= 0:
        raise ValueError("max_range must be positive")
    if not 0 <= obstacle_distance <= max_range:
        raise ValueError("Obstacle distance is out of range")

    readings = np.full(number_of_rays, max_range)
    centre_ray = number_of_rays // 2
    readings[centre_ray] = obstacle_distance

    return readings

distances = simulate_lidar_readings()

print("Simulated LiDAR distances (metres):")
print(distances)
print("Closest obstacle:", np.min(distances), "metres")
```

```python

import numpy as np


def fuse_sensor_data(lidar_distances, camera_obstacles):
    lidar = np.asarray(lidar_distances, dtype=float)

    # Remove invalid and non-positive LiDAR readings.
    valid_lidar = lidar[np.isfinite(lidar) & (lidar > 0)]

    closest_distance = (
        float(np.min(valid_lidar))
        if valid_lidar.size > 0 else None
    )

    # Combine the information from both sensors.
    fused_data = {
        "lidar_min_distance_m": closest_distance,
        "camera_obstacles": camera_obstacles,
        "nearby_obstacle": (
            closest_distance < 1.0
            if closest_distance is not None else None
        )
    }

    return fused_data


# Example sensor readings for testing
example_lidar = [2.5, 1.8, 0.7, float("inf")]

example_camera = [
    {"label": "obstacle", "confidence": 0.85}
]

result = fuse_sensor_data(example_lidar, example_camera)

print("Fused sensor data:")
print(result)
```
